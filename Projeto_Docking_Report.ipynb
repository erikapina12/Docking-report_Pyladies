```python
# Importando os arquivos para análise

# Carregando/instalando os pacotes necessários
!pip install biopython
import ipywidgets as widgets
from IPython.display import display
import io
from Bio.PDB import PDBParser, MMCIFParser


# Habilita a seleção de múltiplos arquivos
uploader = widgets.FileUpload(
    description='Carregar PDBs/CIFs',
    accept='.pdb,.cif,.mmcif',
    multiple=True
)

def on_upload_change(change):
    if not uploader.value:
        return

    # Dicionário para guardar todas as estruturas carregadas
    estruturas_carregadas = {}

    # Itera sobre todos os arquivos enviados pelo usuário
    for uid, file_info in uploader.value.items():
        filename = file_info['metadata']['name']
        content = file_info['content']

        # Salva temporariamente para o Biopython ler
        with open(filename, 'wb') as f:
            f.write(content)

        # Escolhe o parser correto
        if filename.lower().endswith('.pdb'):
            parser = PDBParser(QUIET=True)
        else:
            parser = MMCIFParser(QUIET=True)

        # Carrega e armazena no dicionário usando o nome do arquivo como chave
        estrutura = parser.get_structure(filename, filename)
        estruturas_carregadas[filename] = estrutura
        print(f"Sucesso: '{filename}' carregado!")

    print(f"\nTotal de estruturas carregadas nesta sessão: {len(estruturas_carregadas)}")
    # Agora você pode acessar, por exemplo: estruturas_carregadas['1azm.pdb']

uploader.observe(on_upload_change, names='value')
display(uploader)
```
